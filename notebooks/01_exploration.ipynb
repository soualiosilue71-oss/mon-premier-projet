```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

df = pd.DataFrame({
    "mois": ["Jan", "Fév", "Mar", "Avr", "Mai"],
    "ventes": np.random.randint(100, 200, 5),
})
df
```

```python
df.plot(x="mois", y="ventes", kind="bar", title="Ventes par mois")
plt.savefig("../reports/figures/ventes.png", bbox_inches="tight")
plt.show()
```
